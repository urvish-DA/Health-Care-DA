```python
def healthcare_pipeline():
    import pandas as pd
    base_path = r"C:\Data analyst\healthcare_clinical_analytics_project\healthcare_clinical_analytics_project\02_data_files"
    # Load data
    patients = pd.read_csv(base_path + "\\patients.csv")
    treatments = pd.read_csv(base_path + "\\treatments.csv")
    trial_results = pd.read_csv(base_path + "\\trial_results.csv")
    clinical_trials = pd.read_csv(base_path + "\\clinical_trials.csv")
    doctors = pd.read_csv(base_path + "\\doctors.csv")
    hospitals = pd.read_csv(base_path + "\\hospitals.csv")
    medications = pd.read_csv(base_path + "\\medications.csv")
    # Clean data
    patients["city"] = patients["city"].fillna("Unknown")
    patients["age"] = patients["age"].fillna(patients["age"].median())
    trial_results["side_effect"] = trial_results["side_effect"].fillna("Unknown")
    trial_results["patient_satisfaction_score"] = trial_results["patient_satisfaction_score"].fillna(
        trial_results["patient_satisfaction_score"].mean()
    )

    # Feature engineering
    patients["full_name"] = patients["first_name"] + " " + patients["last_name"]

    # Merge data
    df = pd.merge(patients, trial_results, on="patient_id", how="inner")
    df = pd.merge(df, clinical_trials, on="trial_id", how="inner")
    df = pd.merge(df, hospitals, on="hospital_id", how="inner")

    # Export data
    df.to_csv("f_healthcare_analysis.csv", index=False)

    return df
```

```python
final_df=healthcare_pipeline()
```
